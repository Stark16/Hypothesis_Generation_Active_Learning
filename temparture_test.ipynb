> ## Conclusion 1 - With `chat history=True` I get **different** response for the same input the **2nd time**. 
>> ### I think this is because when `chat history=True`,
>>> ### That means when we run the 1st prompt our input is-
>>>> $input_1 = prompt_1$
>>> ### But due to `context_history=True`, the 2nd input becomes-
>>>> $input_2$ = $prompt_1$ $+$ $response_1$ $+$ $prompt_2$
>>>>>where, $(prompt_1 + response_1) =$ `context history`
>> ### Meaning, the input is no longer same since $input_2 \neq input_1$ even though the prompt in both inputs is $prompt_1$
